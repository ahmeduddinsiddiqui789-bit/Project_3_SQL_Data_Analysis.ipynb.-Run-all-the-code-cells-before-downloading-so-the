```python
import pandas as pd
import sqlite3
from google.colab import files

uploaded = files.upload()

file_name = next(iter(uploaded))
df = pd.read_excel(file_name)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Convert the date into a format SQLite can read
df["Date"] = pd.to_datetime(df["Date"]).dt.strftime("%Y-%m-%d")

# Create the SQLite database
conn = sqlite3.connect("project3.db")

# Store the dataset as a SQL table named orders
df.to_sql("orders", conn, if_exists="replace", index=False)

print("SQL database created successfully!")
print("Table name: orders")
```

```python
query = """
SELECT Product, COUNT(*) AS NumberOfOrders
FROM orders
GROUP BY Product
ORDER BY NumberOfOrders DESC;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT Product,
       SUM(TotalPrice) AS TotalSales
FROM orders
GROUP BY Product
ORDER BY TotalSales DESC;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT Product,
       ROUND(AVG(TotalPrice), 2) AS AverageOrderValue
FROM orders
GROUP BY Product
ORDER BY AverageOrderValue DESC;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT PaymentMethod,
       COUNT(*) AS NumberOfOrders,
       SUM(TotalPrice) AS TotalSales,
       ROUND(AVG(TotalPrice), 2) AS AverageOrderValue
FROM orders
GROUP BY PaymentMethod
ORDER BY NumberOfOrders DESC;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT OrderStatus,
       COUNT(*) AS NumberOfOrders,
       SUM(TotalPrice) AS TotalOrderValue
FROM orders
GROUP BY OrderStatus
ORDER BY NumberOfOrders DESC;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT ReferralSource,
       COUNT(*) AS NumberOfOrders
FROM orders
WHERE ReferralSource = 'Instagram'
GROUP BY ReferralSource;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT Product,
       SUM(TotalPrice) AS TotalSales
FROM orders
GROUP BY Product
HAVING SUM(TotalPrice) > 180000
ORDER BY TotalSales DESC;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT strftime('%Y-%m', Date) AS OrderMonth,
       COUNT(*) AS NumberOfOrders,
       SUM(TotalPrice) AS TotalOrderValue
FROM orders
GROUP BY strftime('%Y-%m', Date)
ORDER BY OrderMonth;
"""

display(pd.read_sql_query(query, conn))
```

```python
sql_code = """
-- Project 3: SQL Data Analysis

-- 1. Display selected columns
SELECT OrderID, Product, Quantity, TotalPrice
FROM orders;

-- 2. Filter orders
SELECT OrderID, Product, TotalPrice
FROM orders
WHERE TotalPrice > 1000;

-- 3. Highest-priced orders
SELECT OrderID, Product, TotalPrice
FROM orders
ORDER BY TotalPrice DESC
LIMIT 10;

-- 4. Count orders by product
SELECT Product, COUNT(*) AS NumberOfOrders
FROM orders
GROUP BY Product;

-- 5. Total sales by product
SELECT Product, SUM(TotalPrice) AS TotalSales
FROM orders
GROUP BY Product;

-- 6. Average order value
SELECT Product, ROUND(AVG(TotalPrice), 2) AS AverageOrderValue
FROM orders
GROUP BY Product;

-- 7. Payment method analysis
SELECT PaymentMethod, COUNT(*) AS NumberOfOrders,
       SUM(TotalPrice) AS TotalSales
FROM orders
GROUP BY PaymentMethod;

-- 8. Order status analysis
SELECT OrderStatus, COUNT(*) AS NumberOfOrders
FROM orders
GROUP BY OrderStatus;

-- 9. Referral source analysis
SELECT ReferralSource, COUNT(*) AS NumberOfOrders
FROM orders
GROUP BY ReferralSource;

-- 10. Products above a sales threshold
SELECT Product, SUM(TotalPrice) AS TotalSales
FROM orders
GROUP BY Product
HAVING SUM(TotalPrice) > 180000;

-- 11. Monthly order analysis
SELECT strftime('%Y-%m', Date) AS OrderMonth,
       COUNT(*) AS NumberOfOrders,
       SUM(TotalPrice) AS TotalOrderValue
FROM orders
GROUP BY strftime('%Y-%m', Date)
ORDER BY OrderMonth;
"""

with open("project3_queries.sql", "w", encoding="utf-8") as f:
    f.write(sql_code)

print("SQL file created successfully!")
files.download("project3_queries.sql")
```
